# **CLEANING**

```python
df.isnull().sum()
```

Terlihat beberapa atribut masih memiliki banyak nilai null (nilai kosong). Seperti driector, duration_minutes, country, cast,date_added, dan rating

```python
df['director'] = (df['director']
                  .fillna('Director not specified')
                  .apply(lambda x: ', '.join([s.strip() for s in str(x).split(',')])
                  if isinstance(x, str) else 'Director not specified')
                  .astype('string'))


df['cast'] = (df['cast']
              .fillna('No cast listed')
              .apply(lambda x: 'No cast listed' if x == 'No cast listed'
                     else ', '.join([s.strip() for s in str(x).split(',')]))
              .astype('string'))


df['country'] = (df['country']
                 .fillna('Country not specified')
                 .apply(lambda x: 'Country not specified' if x == 'Country not specified'
                        else ', '.join([s.strip() for s in str(x).split(',')]))
                 .astype('string'))
```

Membersihkan dan merapikan data pada kolom director, cast, dan country agar lebih konsisten. Nilai yang kosong (NaN) diisi dengan “Director not specified”, “No cast listed”, dan “Country not specified” sehingga tidak ada data yang hilang. Teks pada setiap kolom diproses dengan memisahkan berdasarkan koma, menghapus spasi berlebih pada setiap elemen, lalu menggabungkannya kembali agar formatnya lebih rapi. Terakhir, tipe data diubah menjadi string untuk memastikan keseragaman.

```python
df.isnull().sum()
```

Setelah dilakukan pembersihan ternyata ada beberapa atribut yang masih memiliki nilai null

```python
# Konversi sederhana ke string
df['type'] = df['type'].astype('string')
df['description'] = df['description'].astype('string')
df['listed_in'] = df['listed_in'].astype('string')
```

Dilakukan perubahan type data menjadi string yaitu atribut description dan listed_in

```python
df.info()
```

Setelah dirubah type data, masih ada beberapa atribut yang masih memiliki nilai null seperti date_added, rating, dan duration_minutes

```python
# Konversi date_added ke datetime
df['date_added'] = pd.to_datetime(df['date_added'])
# Isi NaN dengan tanggal sebelumnya (forward fill)
df['date_added'] = df['date_added'].fillna(method='ffill')
# Kalau masih ada NaN di awal isi dengan tanggal sesudahnya (backward fill)
df['date_added'] = df['date_added'].fillna(method='bfill')
```

Dilakukan pengubahan kolom date_added ke format tanggal, lalu mengisi nilai kosong dengan tanggal terdekat menggunakan forward fill dan backward fill agar tidak ada data yang hilang.

```python
if 'duration_minutes' in df.columns:
    median_durasi = df['duration_minutes'].median()
    df['duration_minutes'] = df['duration_minutes'].fillna(median_durasi)
```

Dilakukan pengecekan apakah kolom duration_minutes ada, lalu mengisi nilai yang kosong dengan nilai median dari kolom tersebut agar data tetap representatif tanpa dipengaruhi oleh nilai ekstrem

```python
df.info()
```

Seluruh data sudah tidak ada nilai null kecuali pada atribut rating yang masih memiliki 3 data nilai null

```python
df
```

```python
import numpy as np

rating_to_float = {
    'TV-Y': 1.0, 'TV-Y7': 2.0, 'TV-Y7-FV': 2.5, 'TV-G': 3.0, 'G': 3.0,
    'TV-PG': 4.0, 'PG': 4.0, 'PG-13': 5.0, 'TV-14': 6.0, 'R': 7.0,
    'TV-MA': 7.0, 'NC-17': 8.0
}

modus_rating = df['rating'].mode().iloc[0]
df['rating'] = df['rating'].fillna(modus_rating)

df['rating'] = df['rating'].map(rating_to_float)

median_val = df['rating'].median()
df['rating'] = df['rating'].fillna(median_val).astype(float)

print("Modus rating yang dipakai untuk isi NaN:", modus_rating)
print(df['rating'].value_counts().sort_index())
print(df[['rating']].sample(5))
```

Dilakukan pengubahan data rating dari kategori menjadi angka agar bisa dianalisis secara numerik. Nilai kosong terlebih dahulu diisi dengan modus (rating yang paling sering muncul, yaitu TV-G), kemudian setiap rating dipetakan ke skala angka. Jika setelah pemetaan masih ada nilai kosong, diisi dengan nilai median. Hasilnya, distribusi menunjukkan bahwa rating 3.0 (TV-G/G) dan 4.0 (TV-PG/PG) paling dominan, yang mengindikasikan mayoritas konten bersifat ramah keluarga.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df['rating_std'] = scaler.fit_transform(df[['rating']])

print(df[['rating', 'rating_std']].head())
```

melakukan standarisasi (Z-score) pada kolom rating menggunakan StandardScaler, sehingga nilai diubah memiliki rata-rata 0 dan standar deviasi 1. Hasilnya, nilai seperti 3.0 menjadi negatif (di bawah rata-rata) dan 4.0 menjadi positif (di atas rata-rata), sehingga memudahkan analisis dan pemodelan karena skala data sudah seragam.

```python
from sklearn.preprocessing import MinMaxScaler

minmax_scaler = MinMaxScaler()

df['rating_norm'] = minmax_scaler.fit_transform(df[['rating']])

df['release_year_norm'] = minmax_scaler.fit_transform(df[['release_year']])

print(df[['rating', 'rating_norm', 'release_year', 'release_year_norm']].head())
```

melakukan normalisasi Min-Max pada kolom rating dan release_year sehingga nilainya berada dalam rentang 0 hingga 1. Hasilnya, nilai rating seperti 3.0 menjadi 0.4 dan 4.0 menjadi 0.6, sementara tahun rilis yang lebih baru memiliki nilai mendekati 1. Proses ini membuat perbandingan antar fitur lebih seimbang dan memudahkan penggunaan dalam model analisis atau machine learning.

```python
from sklearn.preprocessing import LabelEncoder

le_type = LabelEncoder()
df['type_encoded'] = le_type.fit_transform(df['type'])
mapping_type = pd.DataFrame({
    'Original Type': le_type.classes_,
    'Encoded Value': le_type.transform(le_type.classes_)
})

le_rating = LabelEncoder()
df['rating_encoded'] = le_rating.fit_transform(df['rating'])
mapping_rating = pd.DataFrame({
    'Original Rating': le_rating.classes_,
    'Encoded Value': le_rating.transform(le_rating.classes_)
})

print("\n=== Mapping 'type' ===")
print(mapping_type.to_string(index=False))

print("\n=== Mapping 'rating' ===")
print(mapping_rating.to_string(index=False))
```

proses encoding data kategorikal menjadi numerik. Pada kolom type, nilai Movie diubah menjadi 0 dan TV Show menjadi 1 agar bisa diproses oleh model. Sementara itu, pada rating, nilai yang sebelumnya sudah berbentuk angka (hasil konversi) diubah lagi menjadi label berurutan dari 0 hingga 6. Insight-nya, encoding ini membantu menyederhanakan representasi data sehingga lebih mudah digunakan dalam analisis atau algoritma machine learning, meskipun urutan angka juga bisa memberi kesan adanya tingkatan antar rating.

```python
# --- One-Hot Encoding untuk kolom 'country' ---
# Ambil 5 negara dengan frekuensi terbanyak
top_countries = df['country'].value_counts().head(5).index

# Buat series baru dengan hanya negara teratas, sisanya jadi NaN
country_filtered = df['country'].where(df['country'].isin(top_countries))

# Buat dummy variabel untuk negara teratas
df_country_dummies = pd.get_dummies(country_filtered, prefix='country')

# Gabungkan dummy negara ke dataframe utama
df = pd.concat([df, df_country_dummies], axis=1)


# --- One-Hot Encoding untuk genre di kolom 'listed_in' ---
# Pisahkan genre yang digabung dengan koma, buat dummy variabel
genre_dummies = df['listed_in'].str.get_dummies(sep=', ')

# Gabungkan dummy genre ke dataframe utama
df = pd.concat([df, genre_dummies], axis=1)


# --- Opsional: lihat hasil dummy negara dan genre ---
print("Dummy negara (top 5):")
print(df_country_dummies.head())

print("\nDummy genre:")
print(genre_dummies.head())

print("\nBeberapa kolom hasil encoding pada dataframe utama:")
print(df[['country', 'listed_in'] + list(df_country_dummies.columns) + list(genre_dummies.columns)].head())
```

Hasil tersebut menunjukkan proses one-hot encoding (dummy variables) pada kolom country dan listed_in (genre), di mana setiap kategori diubah menjadi kolom biner (0/1 atau True/False). Pada country, hanya diambil beberapa negara utama (top 5) sehingga setiap baris menunjukkan asal negara dalam bentuk kolom terpisah. Sementara itu, pada genre, setiap jenis genre direpresentasikan sebagai kolom sendiri dengan nilai 1 jika konten memiliki genre tersebut dan 0 jika tidak. Teknik ini memungkinkan model memahami data kategorikal tanpa mengasumsikan urutan, sekaligus menangkap bahwa satu konten bisa memiliki lebih dari satu genre atau kombinasi negara.

# **JOINING**

```python
df_valid_director = df[df['director'] != 'Director not specified'].copy()

df_copy = df_valid_director[['title', 'director']].copy()
df_copy = df_copy.rename(columns={'title': 'other_title'})

df_joined = pd.merge(df_valid_director, df_copy, on='director')

df_joined = df_joined[df_joined['title'] != df_joined['other_title']]

print(df_joined[['title', 'other_title', 'director']].head())
```

digunakan untuk menemukan keterkaitan antar konten berdasarkan sutradara yang sama. Data terlebih dahulu difilter agar hanya mencakup entri dengan sutradara yang valid, kemudian dilakukan self-join pada kolom director untuk memasangkan setiap judul dengan judul lain yang disutradarai oleh orang yang sama. Baris dengan judul yang sama dihapus agar hanya tersisa pasangan konten yang berbeda. Hasil ini membantu mengidentifikasi karya-karya lain dari sutradara yang sama, sehingga bisa digunakan untuk melihat pola produksi atau rekomendasi konten yang berkaitan.

```python
# Baris yang memiliki other_title
df_joined_with_other = df_joined[df_joined['other_title'].notnull()]
print(df_joined_with_other.head())
```

setiap baris menunjukkan satu judul beserta judul lain yang terkait melalui sutradara yang sama. Data ini bisa digunakan untuk melihat hubungan antar film (misalnya karya dari sutradara yang sama) sehingga berguna untuk analisis pola produksi atau rekomendasi konten yang serupa.

```python
import os

output_dir = './output'

os.makedirs(output_dir, exist_ok=True)

df_final = pd.merge(df, df_joined[['title', 'other_title']], on='title', how='left')

df_final.to_csv(os.path.join(output_dir, 'disney_plus_titles_final.csv'), index=False)

print("Isi folder output:")
print(os.listdir(output_dir))
```

```python
kids_keywords = ['Kids', 'Family']
kids_ratings = ['TV-Y', 'TV-Y7', 'G', 'PG']

df['is_kids'] = df.apply(
    lambda row: 1 if any(kw.lower() in str(row['listed_in']).lower() for kw in kids_keywords) or
                     any(rating in str(row['rating']) for rating in kids_ratings)
                else 0,
    axis=1
)

print(df[['title', 'listed_in', 'rating', 'is_kids']].head())
```

membuat fitur baru is_kids untuk mengidentifikasi apakah suatu konten cocok untuk anak-anak atau tidak. Penentuan dilakukan berdasarkan dua hal, yaitu apakah genre (listed_in) mengandung kata seperti Kids atau Family, atau apakah rating termasuk kategori ramah anak seperti TV-Y, TV-Y7, G, dan PG. Jika salah satu kondisi terpenuhi, maka diberi nilai 1 (konten anak), jika tidak maka 0. Insight-nya, fitur ini membantu mengelompokkan konten secara sederhana berdasarkan target usia sehingga bisa digunakan untuk analisis atau sistem rekomendasi khusus anak.

# **RESAMPLING**

```python
df = df.set_index('date_added')

monthly_total = df['show_id'].resample('M').count()

monthly_series = df[df['type'] == 'TV Show']['show_id'].resample('M').count()
monthly_movies = df[df['type'] == 'Movie']['show_id'].resample('M').count()

def extract_duration(x):
    import re
    if pd.isna(x):
        return None
    match = re.search(r'(\d+)', str(x))
    if match:
        return int(match.group(1))
    return None

df['duration_num'] = df['duration'].apply(extract_duration)

monthly_avg_duration = df['duration_num'].resample('M').mean()

print("Jumlah tayangan per bulan:")
print(monthly_total.head())
print("\nJumlah series per bulan:")
print(monthly_series.head())
print("\nJumlah movies per bulan:")
print(monthly_movies.head())

print("\nRata-rata durasi tayangan per bulan (menit/seasons):")
print(monthly_avg_duration.head())
```

digunakan untuk menganalisis tren penambahan konten per bulan. Data diatur berdasarkan date_added, lalu dihitung jumlah total tayangan, serta dipisahkan antara TV Show dan Movie. Hasilnya menunjukkan lonjakan besar pada November 2019, yang mengindikasikan adanya penambahan konten secara massal pada periode tersebut, dengan film sebagai kontributor utama. Selain itu, durasi diubah ke bentuk numerik untuk menghitung rata-rata durasi per bulan, yang terlihat cukup bervariasi—menandakan campuran antara konten berdurasi pendek dan panjang. Pola ini bisa mencerminkan strategi platform dalam merilis konten, seperti peluncuran awal atau penambahan katalog secara besar-besaran di waktu tertentu.
